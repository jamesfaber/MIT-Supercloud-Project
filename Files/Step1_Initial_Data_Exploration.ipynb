```python
# Import Libraries and Connect to AWS S3
import pandas as pd
pd.set_option('display.max_columns', None)
import s3fs
fs = s3fs.S3FileSystem(anon=True)
```

```python
# List the top-level directories of the MIT Supercloud dataset
print(*fs.ls('s3://mit-supercloud-dataset/'), sep='\n')
```

```python
# List the directories within the datacenter challenge directory
print(*fs.ls('s3://mit-supercloud-dataset/datacenter-challenge/'), sep='\n')
```

```python
# List the directories within the 202201 directory
print(*fs.ls('s3://mit-supercloud-dataset/datacenter-challenge/202201/'), sep='\n')
```
